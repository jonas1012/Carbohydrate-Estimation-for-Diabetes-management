This script is pretty basic it will calculate the PMAE for select models, in order to more easily compare them to comparative models. PMAE is simply calculated by MAE / mean ground truth value * 100. since every dataset was tested on the same test set will we just need to input the MAE value, and then we are able to calculate the PMAE values.

```python
import pandas as pd
import numpy as np

#Loading metadata

TEST_METADATA_PATH = "/content/drive/MyDrive/Speciale/dataset/prepared_overhead_rgb_dataset/dish_carb_mass_ingredients_split_metadata.csv"

df = pd.read_csv(TEST_METADATA_PATH)

# Keep only test set
test_df = df[df["split"] == "test"].copy()

print(test_df.columns)

CARB_COL = "total_carb"

mean_test_carb = test_df[CARB_COL].mean()

print(f"Mean carbohydrate value in test set: {mean_test_carb:.4f} g")

#Model MAEs

model_results = {
    "Model 3 V3": 4.7653,
    "Model 22": 4.6187,
    "Model 24": 4.7191,
    "Model 11": 4.0446,
}

#Calculate PMAE

rows = []

for model_name, mae in model_results.items():
    pmae = (mae / mean_test_carb) * 100

    rows.append({
        "Model": model_name,
        "MAE": mae,
        "Mean test carbs": mean_test_carb,
        "PMAE (%)": pmae
    })

pmae_table = pd.DataFrame(rows)

pmae_table["MAE"] = pmae_table["MAE"].round(4)
pmae_table["Mean test carbs"] = pmae_table["Mean test carbs"].round(4)
pmae_table["PMAE (%)"] = pmae_table["PMAE (%)"].round(2)

display(pmae_table)
```
