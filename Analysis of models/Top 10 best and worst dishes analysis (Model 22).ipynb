This script will calculate the three top 10 best and worst dishes for model 22 Three tables will be presented with absolute error for the first, percentage error for the second and percentage error after removal of dishes with a ground truth value of under 5g.

```python
from pathlib import Path
import numpy as np
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 180)

MODEL22_DIR = Path("/content/drive/MyDrive/Speciale/Models/Model_22_Carbohydrate_estimator_plus_ingredient_estimator")

fixed_csv_path = MODEL22_DIR / "model22_test_predictions_FIXED.csv"

df = pd.read_csv(fixed_csv_path)

print("Loaded fixed prediction CSV:")
print(fixed_csv_path)
print("Shape:", df.shape)

display(df.head())
print(df.columns.tolist())
```

Loading prediction CSV

```python
target_col = "total_carb"
rgb_col = "rgb_only_pred"
model22_col = "model22_pred"

required_cols = ["dish_id", target_col, rgb_col, model22_col]

missing_cols = [col for col in required_cols if col not in df.columns]

if missing_cols:
    raise ValueError(f"Missing columns in CSV: {missing_cols}")

# Absolute errors
df["rgb_abs_error"] = (df[rgb_col] - df[target_col]).abs()
df["model22_abs_error"] = (df[model22_col] - df[target_col]).abs()

# Error improvement: positive means Model 22 improved compared to RGB-only / Model 3 V3
df["abs_error_improvement"] = df["rgb_abs_error"] - df["model22_abs_error"]

# Percentage errors
df["rgb_percentage_error"] = np.where(
    df[target_col] > 0,
    (df["rgb_abs_error"] / df[target_col]) * 100,
    np.nan
)

df["model22_percentage_error"] = np.where(
    df[target_col] > 0,
    (df["model22_abs_error"] / df[target_col]) * 100,
    np.nan
)

# Percentage error improvement
df["percentage_error_improvement"] = df["rgb_percentage_error"] - df["model22_percentage_error"]

display_cols = [
    "dish_id",
    target_col,
    rgb_col,
    model22_col,
    "rgb_abs_error",
    "model22_abs_error",
    "abs_error_improvement",
    "rgb_percentage_error",
    "model22_percentage_error",
    "percentage_error_improvement",
]

display(df[display_cols].head())
```

```python
def create_top10_table(data, ranking_col, title):
    print("=" * 120)
    print(title)
    print("=" * 120)

    table_cols = [
        "dish_id",
        target_col,
        rgb_col,
        model22_col,
        "rgb_abs_error",
        "model22_abs_error",
        "abs_error_improvement",
        "rgb_percentage_error",
        "model22_percentage_error",
        "percentage_error_improvement",
    ]

    best_10 = data.sort_values(ranking_col, ascending=True).head(10).copy()
    worst_10 = data.sort_values(ranking_col, ascending=False).head(10).copy()

    print("\nTop 10 best Model 22 predictions:")
    display(best_10[table_cols].round(2))

    print("\nTop 10 worst Model 22 predictions:")
    display(worst_10[table_cols].round(2))

    return best_10, worst_10
```

Version 1 Top 10 by absolute error

```python
best_abs, worst_abs = create_top10_table(
    data=df,
    ranking_col="model22_abs_error",
    title="Top 10 best and worst Model 22 predictions based on absolute error"
)
```

Version 2 Top 10 by percentage error

```python
df_pct = df.dropna(subset=["model22_percentage_error"]).copy()

best_pct, worst_pct = create_top10_table(
    data=df_pct,
    ranking_col="model22_percentage_error",
    title="Top 10 best and worst Model 22 predictions based on percentage error"
)
```

Version 3 Top 10 by percentage error after removing dishes under 5g carbohydrates

```python
MIN_CARB_THRESHOLD = 5.0

df_pct_min5 = df_pct[df_pct[target_col] >= MIN_CARB_THRESHOLD].copy()

print("Original test set size:", len(df))
print("Test set size after removing dishes below 5g carbohydrates:", len(df_pct_min5))
print("Removed dishes:", len(df) - len(df_pct_min5))

best_pct_min5, worst_pct_min5 = create_top10_table(
    data=df_pct_min5,
    ranking_col="model22_percentage_error",
    title="Top 10 best and worst Model 22 predictions based on percentage error, excluding dishes below 5g carbohydrates"
)
```
